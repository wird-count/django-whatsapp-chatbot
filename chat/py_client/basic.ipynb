```python
import requests
# endpoint = 'https:///www.github.com' 
endpoint = 'https://httpbin.org/cookies'
endpoint = 'https://httpbin.org/status/200'
endpoint = 'https://httpbin.org/'
endpoint = 'https://httpbin.org/anything' 
# endpoint  = 'http://localhost:8000/'

def lancee_une_requette(endpoint):
    get_response = requests.get(endpoint,json={'user':'W9'},data=f'no data now ya monsieur') #HTTP requests donc quand on envoie une request manuelle evec python rest api (contrairement a automatic avec postman) ici la request elle a des reponse donc que on devrais exploitez dailleur car cest tout a fait le principe des cette sychronization backend-backend 
    print(get_response.text)
    print(get_response.status_code)
    print('_______')
    print(get_response.json())
    print('_______')
    print(get_response.headers)
    print('_______')
# remarc :
# http request --> HTML
# rest api http request --> json

lancee_une_requette(endpoint)


```
